```python
import nltk
nltk.download('punkt_tab')
```

```python
from nltk import ngrams
```

```python
from nltk.tokenize import word_tokenize,sent_tokenize
```

```python
sent="Learning nlp is really  interesting so come to class regular "
```

```python
words=word_tokenize(sent)
```

```python
print("tokens",words)
```

**unigrams**

```python
unigrams = list(ngrams(words, 1))
print("\nUnigrams:")
print(unigrams)
```

**Bigrams**

```python
bigrams = list(ngrams(words, 2))
print("\nBigrams:")
print(bigrams)
```

**trigrams**

```python
trigrams = list(ngrams(words, 3))
print("\nTrigrams:")
print(trigrams)
```
